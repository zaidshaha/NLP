```python
text = """
Once upon a time, there lived a young girl in a small village. She loved reading books and playing with her friends. 
Every morning, she woke up early and went to school. Her teachers were kind, and they taught the students many interesting things. 
One day, she found a beautiful bird sitting under a tree. The bird looked weak, so she gently picked it up and gave it some food and water. 
After a few days, the bird became healthy and flew happily into the sky.
"""
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print("Text processed successfully.")
print("Number of tokens:", len(doc))
```

```python
for token in doc:
    print(token.text, "->", token.pos_)
```

```python
import pandas as pd

pos_table = pd.DataFrame({
"Token": [token.text for token in doc],
"POS": [token.pos_ for token in doc]
})
print(pos_table.to_string(index=True))
```

```python
detailed_table = pd.DataFrame({
"Token": [token.text for token in doc],
"POS": [token.pos_ for token in doc],
"Tag": [token.tag_ for token in doc]
})

print(detailed_table.to_string(index=True))
```

```python
analysis_table = pd.DataFrame({
"Token": [token.text for token in doc],
"POS": [token.pos_ for token in doc],
"Tag": [token.tag_ for token in doc],
"Lemma": [token.lemma_ for token in doc],
"Morphology": [str(token.morph) for token in doc]
})

print(analysis_table.to_string(index=True))
```

```python
word_table = pd.DataFrame({
"Token": [t.text for t in doc if t.is_alpha],
"POS": [t.pos_ for t in doc if t.is_alpha],
"Tag": [t.tag_ for t in doc if t.is_alpha],
"Lemma": [t.lemma_ for t in doc if t.is_alpha]
})

print(word_table.to_string(index=True))
```

```python
pos_counts = (
word_table["POS"]
.value_counts()
.rename_axis("POS")
.reset_index(name="Count")
)

print(pos_counts)
```

```python
import matplotlib.pyplot as plt

pos_counts.plot(
kind="bar",
x="POS",
y="Count",
legend=False,
title="POS Tag Distribution"
)

plt.xlabel("Part of Speech")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

1. Case Study 1: Biomedical & Chemical Parsing 

```python
import re
from nltk.tokenize import TreebankWordTokenizer
text = "Ca2+ concentrations at pH 7.4 (in vitro)."
print("=" * 65)
print("CASE STUDY 1: BIOMEDICAL & CHEMICAL PARSING")
print("Input:", text)
print("=" * 65)

ws_tokens = text.split()
print("\n[1] White-Space Tokenization:")
print(ws_tokens)

naive_punct_tokens = [tok for tok in re.split(r"\W+", text) if tok]
print("\n[2] Naive Non-Word Split (\\W+):")
print(naive_punct_tokens)

ptb_tokenizer = TreebankWordTokenizer()
rule_aware_tokens = ptb_tokenizer.tokenize(text)
print("\n[3] Rule-Aware Tokenization (TreebankWordTokenizer):")
print(rule_aware_tokens)
```

Case Study 2: Social Media & Web Noise

```python
import re
from nltk.tokenize import TweetTokenizer, word_tokenize
tweet = "@user Check https://site.org #NLP :-)"
print("=" * 65)
print("CASE STUDY 2: SOCIAL MEDIA & WEB NOISE")
print("Input:", tweet)
print("=" * 65)

std_tokens = word_tokenize(tweet)
print("\n[1] Standard Word Tokenizer:")
print(std_tokens)

naive_tokens = [tok for tok in re.split(r"[\s]+", tweet) if tok]
print("\n[2] Naive Whitespace Split:")
print(naive_tokens)

tweet_tok = TweetTokenizer(preserve_case=True, reduce_len=True, strip_handles=False)
domain_tokens = tweet_tok.tokenize(tweet)
print("\n[3] Domain-Aware Tokenizer (TweetTokenizer):")
print(domain_tokens)

```

Tokenization Error Propagation Pipeline

```python
import nltk
from nltk import pos_tag
print("=" * 65)
print("PIPELINE ERROR PROPAGATION (SLIDES 6-9)")
print("Sentence: 'we\'re ready'")
print("=" * 65)

naive_tokens = ["we're", "ready"]
naive_pos = pos_tag(naive_tokens)
print("\n[Scenario A] Naive Tokenization & POS Tagging:")
print("Tokens :", naive_tokens)
print("POS Tags:", naive_pos)


ptb_tokens = ["we", "'re", "ready"]
correct_pos = pos_tag(ptb_tokens)
print("\n[Scenario B] Rule-Aware Tokenization & POS Tagging:")
print("Tokens :", ptb_tokens)
print("POS Tags:", correct_pos)
```
