Import Libraries

```python
import spacy
import nltk
import pandas as pd

from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer
```

```python
nlp = spacy.load("en_core_web_sm")

print("spaCy version:", spacy.__version__)
print("NLTK version:", nltk.__version__)
print("English model loaded successfully!")
```

Input

```python
text = """
Online shoppers are searching for affordable products.
Customers compare prices, read reviews, and purchase useful items.
Many shoppers were buying products online because stores offered convenient services.
The companies are improving their websites and providing better recommendations.
"""

print(text)
print("Character count:", len(text))
```

Create spaCy Document

```python
doc = nlp(text)

print("Number of spaCy tokens:", len(doc))
```

Tokenization

A. Whitespace Tokenization

```python
whitespace_tokens = text.split()

print("Whitespace Tokens:")
print(whitespace_tokens)

print("\nNumber of tokens:", len(whitespace_tokens))
```

Observation

Whitespace tokenization simply splits wherever there is a space.

spaCy Tokenization

```python
spacy_tokens = [token.text for token in doc]

print("spaCy Tokens:")
print(spacy_tokens)

print("\nNumber of tokens:", len(spacy_tokens))
```

Remove Punctuation and Spaces

```python
clean_tokens = [
    token.text
    for token in doc
    if not token.is_space and not token.is_punct
]

print("Filtered Tokens:")
print(clean_tokens)

print("\nNumber of clean tokens:", len(clean_tokens))
```

Porter Stemming

```python
porter = PorterStemmer()

porter_stems = [
    porter.stem(token.text.lower())
    for token in doc
    if token.is_alpha
]

print("Porter Stems:")
print(porter_stems)
```

```python
print("Original Word -> Porter Stem")
print("-" * 35)

for token in doc:
    if token.is_alpha:
        print(f"{token.text:20} -> {porter.stem(token.text.lower())}")
```

Snowball Stemming

```python
snowball = SnowballStemmer("english")

snowball_stems = [
    snowball.stem(token.text.lower())
    for token in doc
    if token.is_alpha
]

print("Snowball Stems:")
print(snowball_stems)
```

Compare it with Porter:

```python
print("Original Word -> Porter -> Snowball")
print("-" * 50)

for token in doc:
    if token.is_alpha:
        
        word = token.text.lower()
        porter_result = porter.stem(word)
        snowball_result = snowball.stem(word)
        
        print(
            f"{word:20} -> "
            f"{porter_result:15} -> "
            f"{snowball_result}"
        )
```

spaCy Lemmatization

```python
lemmas = [
    token.lemma_.lower()
    for token in doc
    if token.is_alpha
]

print("spaCy Lemmas:")
print(lemmas)
```

Original vs Stem vs Lemma(IMP)

```python
print("Original Word -> Porter Stem -> Snowball Stem -> Lemma")
print("=" * 65)

for token in doc:
    
    if token.is_alpha:
        
        word = token.text.lower()
        porter_result = porter.stem(word)
        snowball_result = snowball.stem(word)
        lemma_result = token.lemma_.lower()
        
        print(
            f"{word:18} "
            f"{porter_result:15} "
            f"{snowball_result:15} "
            f"{lemma_result}"
        )
```

Create Comparison Table

```python
comparison = pd.DataFrame({
    "Original": [
        token.text
        for token in doc
        if token.is_alpha
    ],

    "POS": [
        token.pos_
        for token in doc
        if token.is_alpha
    ],

    "Porter Stem": [
        porter.stem(token.text.lower())
        for token in doc
        if token.is_alpha
    ],

    "Snowball Stem": [
        snowball.stem(token.text.lower())
        for token in doc
        if token.is_alpha
    ],

    "spaCy Lemma": [
        token.lemma_.lower()
        for token in doc
        if token.is_alpha
    ]
})

comparison
```

POS and Lemma

```python
for token in doc:
    
    if token.is_alpha:
        
        print(
            f"Word: {token.text:18} "
            f"POS: {token.pos_:8} "
            f"Lemma: {token.lemma_}"
        )
```

Vocabulary Comparison

```python
original_words = [
    token.text.lower()
    for token in doc
    if token.is_alpha
]

porter_words = [
    porter.stem(word)
    for word in original_words
]

snowball_words = [
    snowball.stem(word)
    for word in original_words
]

lemma_words = [
    token.lemma_.lower()
    for token in doc
    if token.is_alpha
]
```

Vocabulary Reduction Percentage

```python
def reduction_percentage(original, normalized):
    
    original_unique = len(set(original))
    normalized_unique = len(set(normalized))
    
    return (1 - normalized_unique / original_unique) * 100
```

```python
print(
    "Porter vocabulary reduction:",
    round(
        reduction_percentage(original_words, porter_words),
        2
    ),
    "%"
)

print(
    "Snowball vocabulary reduction:",
    round(
        reduction_percentage(original_words, snowball_words),
        2
    ),
    "%"
)

print(
    "spaCy lemma vocabulary reduction:",
    round(
        reduction_percentage(original_words, lemma_words),
        2
    ),
    "%"
)
```

Observations
  
1) Whitespace tokenization divides the text based only on spaces, so punctuation may remain attached to words.
2) spaCy tokenization provides more linguistically meaningful token boundaries and separates punctuation.
3) Porter and Snowball stemming reduce words using heuristic rules.
4) Stemming can produce truncated forms that are not valid English words.
5) spaCy lemmatization produces more interpretable dictionary/base forms.
6) Related word forms such as buying, improving, and providing are normalized differently by stemming and lemmatization.
7) Vocabulary reduction should not be considered the only measure of preprocessing quality.
8) Lemmatization generally provides better semantic coherence than crude stemming.



Conclusion: 

Tokenization, stemming, and lemmatization were successfully performed on the given text using spaCy and NLTK. Whitespace and spaCy tokenization were compared, Porter and Snowball stemming were applied, and spaCy lemmatization was performed. The results demonstrate that stemming provides faster but sometimes truncated word forms, whereas lemmatization generally produces more meaningful linguistic base forms.
